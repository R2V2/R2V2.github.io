```python
import nltk
from nltk.corpus import CHILDESCorpusReader
from collections import defaultdict, Counter

corpus_root = '/Users/hermes/Desktop/coding/corpora/linguistic/super_CHILDES'
prov_root = '/Users/hermes/Desktop/coding/corpora/linguistic/super_CHILDES/providence'
CHILDESbank = CHILDESCorpusReader(corpus_root, '.*xml')
providence = CHILDESCorpusReader(prov_root, '.*xml')
```

```python
corpusAges = providence.age(month=True)
for x in corpusAges:
    print(x)
```

```python
# all child uttterances
child_corpus_counts = defaultdict(Counter)
for word, tag in providence.tagged_words(speaker='CHI'):
    if word in child_corpus_counts:
            child_corpus_counts[word][tag] +=1
    else:
            child_corpus_counts[word] = Counter({tag:1})
```

```python
possibleWords = ["maybe", "can", "have", "hafta", "hasta", "could", "should", "sure", "or", "must", "probably"]
maybe = [sent for sent in providence.sents(speaker="CHI") if "maybe" in sent]
can = [sent for sent in providence.sents(speaker="CHI") if "can" in sent]
hafta = [sent for sent in providence.sents(speaker="CHI") if "hafta" in sent]
hasta = [sent for sent in providence.sents(speaker="CHI") if "hasta" in sent]
have = [sent for sent in providence.sents(speaker="CHI") if "have" in sent]
could = [sent for sent in providence.sents(speaker="CHI") if "could" in sent]
OR = [sent for sent in providence.sents(speaker="CHI") if "or" in sent]
must = [sent for sent in providence.sents(speaker="CHI") if "must" in sent]
sure = [sent for sent in providence.sents(speaker="CHI") if "sure" in sent]
should = [sent for sent in providence.sents(speaker="CHI") if "should" in sent]
probably = [sent for sent in providence.sents(speaker="CHI") if "probably" in sent]
for x in possibleWords:
    print(x, child_corpus_counts[x], sep='\t')
```
